```python
# This is necessary to recognize the modules
import os
import sys
from decimal import Decimal
import warnings

warnings.filterwarnings("ignore")

root_path = os.path.abspath(os.path.join(os.getcwd(), '../..'))
sys.path.append(root_path)
```

```python
from core.backtesting import BacktestingEngine

backtesting = BacktestingEngine(root_path=root_path, load_cached_data=False)
```

```python
from hummingbot.strategy_v2.executors.position_executor.data_types import TrailingStop
from controllers.directional_trading.esp_fib_piv import ESPFibPivConfig
import datetime
from decimal import Decimal


# Controller configuration
connector_name = "binance_perpetual"
trading_pair = "WLD-USDT"
interval = "1m"
total_amount_quote = 1000
max_executors_per_side = 1
take_profit = 0.28
stop_loss = 0.07500000000000001
trailing_stop_activation_price = 0.025
trailing_stop_trailing_delta = 0.018000000000000002
time_limit = 7200  # 2 hours
cooldown_time = 60  # 1 minute
leverage = 20

# ESP Fib Piv specific parameters
ema_fast = 4
ema_medium = 19
ema_slow = 13
atr_length = 13
atr_multiplier = 1.4000000000000001
volume_surge = 1.4
rsi_period = 17
fib_pivot_period = 12
fib_confirmation_threshold = 0.7


# Creating the instance of the configuration and the controller
config = ESPFibPivConfig(
    id=f"esp_fib_piv_{connector_name}_{interval}_{trading_pair}",
    connector_name=connector_name,
    trading_pair=trading_pair,
    interval=interval,
    ema_fast=ema_fast,
    ema_medium=ema_medium,
    ema_slow=ema_slow,
    atr_length=atr_length,
    atr_multiplier=atr_multiplier,
    volume_surge=Decimal(volume_surge),
    rsi_period=rsi_period,
    fib_pivot_period=fib_pivot_period,
    fib_confirmation_threshold=Decimal(fib_confirmation_threshold),
    total_amount_quote=Decimal(total_amount_quote),
    take_profit=Decimal(take_profit),
    stop_loss=Decimal(stop_loss),
    trailing_stop=TrailingStop(
        activation_price=Decimal(trailing_stop_activation_price),
        trailing_delta=Decimal(trailing_stop_trailing_delta)
    ),
    time_limit=time_limit,
    max_executors_per_side=max_executors_per_side,
    cooldown_time=cooldown_time,
)
```

```python
# Running the backtesting this will output a backtesting result object that has built in methods to visualize the results

start = int(datetime.datetime(2024, 9, 1).timestamp())
end = int(datetime.datetime(2025, 3, 1).timestamp())


backtesting_result = await backtesting.run_backtesting(config, start, end, "1m")
```

```python
# Let's see what is inside the backtesting results
print(backtesting_result.get_results_summary())
backtesting_result.get_backtesting_figure()
```

```python
# Display executors on chart with relevant indicators
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

# Get all data from backtesting_result
executors_df = backtesting_result.executors_df

# Get market data from the backtesting_result
# The backtesting_result has market_data directly, not through a controller attribute
candles_df = backtesting_result.market_data.copy()

# Convert timestamps to datetime for better readability
candles_df['timestamp'] = pd.to_datetime(candles_df.index, unit='s')
executors_df['timestamp'] = pd.to_datetime(executors_df['timestamp'], unit='s')

# Create subplots: 3 rows for price/indicators, volume, and RSI
fig = make_subplots(
    rows=3, 
    cols=1, 
    shared_xaxes=True,
    vertical_spacing=0.05,
    subplot_titles=('Price with EMAs, ATR Bands & Trades', 'Volume', 'RSI'),
    row_heights=[0.6, 0.2, 0.2]
)

# Add candlestick chart
fig.add_trace(
    go.Candlestick(
        x=candles_df['timestamp'],
        open=candles_df['open'],
        high=candles_df['high'],
        low=candles_df['low'],
        close=candles_df['close'],
        name='Candlesticks'
    ),
    row=1, col=1
)

# Add EMAs
for ema_name, color in [('ema_fast', 'yellow'), ('ema_medium', 'orange'), ('ema_slow', 'red')]:
    if ema_name in candles_df.columns:
        fig.add_trace(
            go.Scatter(
                x=candles_df['timestamp'],
                y=candles_df[ema_name],
                name=f"{ema_name.replace('_', ' ').upper()} ({getattr(config, ema_name)})",
                line=dict(color=color)
            ),
            row=1, col=1
        )

# Add ATR bands if available
if 'atr_upper_band' in candles_df.columns and 'atr_lower_band' in candles_df.columns:
    fig.add_trace(
        go.Scatter(
            x=candles_df['timestamp'],
            y=candles_df['atr_upper_band'],
            name=f'ATR Upper Band ({config.atr_length}, {config.atr_multiplier})',
            line=dict(color='rgba(173, 216, 230, 0.7)', width=1),
            fill=None
        ),
        row=1, col=1
    )
    
    fig.add_trace(
        go.Scatter(
            x=candles_df['timestamp'],
            y=candles_df['atr_lower_band'],
            name=f'ATR Lower Band ({config.atr_length}, {config.atr_multiplier})',
            line=dict(color='rgba(173, 216, 230, 0.7)', width=1),
            fill='tonexty',
            fillcolor='rgba(173, 216, 230, 0.1)'
        ),
        row=1, col=1
    )

# Add Fibonacci pivot levels if available
fib_columns = [col for col in candles_df.columns if 'fib_' in col and 'threshold' not in col]
for fib_col in fib_columns:
    if fib_col in candles_df.columns:
        fig.add_trace(
            go.Scatter(
                x=candles_df['timestamp'],
                y=candles_df[fib_col],
                name=fib_col.replace('_', ' ').title(),
                line=dict(color='purple', width=1, dash='dot')
            ),
            row=1, col=1
        )

# Add trade entries and exits
for side, color, symbol in [('BUY', 'green', 'triangle-up'), ('SELL', 'red', 'triangle-down')]:
    side_df = executors_df[executors_df['side'] == side]
    
    # Extract entry and exit prices from custom_info
    entry_prices = []
    exit_prices = []
    timestamps = []
    
    for idx, row in side_df.iterrows():
        try:
            custom_info = row['custom_info']
            if isinstance(custom_info, dict):
                entry_price = float(custom_info.get('entry_price', 0))
                close_price = float(custom_info.get('close_price', 0))
                
                if entry_price > 0:
                    entry_prices.append(entry_price)
                    timestamps.append(row['timestamp'])
                
                if close_price > 0:
                    exit_prices.append(close_price)
        except:
            continue
    
    # Add entry markers
    fig.add_trace(
        go.Scatter(
            x=timestamps,
            y=entry_prices,
            mode='markers',
            marker=dict(
                symbol=symbol,
                size=12,
                color=color,
                line=dict(width=1, color='black')
            ),
            name=f'{side} Entry'
        ),
        row=1, col=1
    )
    
    # Add exit markers if available
    if exit_prices and len(timestamps) == len(exit_prices):
        fig.add_trace(
            go.Scatter(
                x=timestamps,
                y=exit_prices,
                mode='markers',
                marker=dict(
                    symbol='circle',
                    size=8,
                    color=color,
                    line=dict(width=1, color='black')
                ),
                name=f'{side} Exit'
            ),
            row=1, col=1
        )

# Add volume chart
fig.add_trace(
    go.Bar(
        x=candles_df['timestamp'],
        y=candles_df['volume'],
        name='Volume',
        marker=dict(color='rgba(100, 100, 255, 0.7)')
    ),
    row=2, col=1
)

# Add volume surge line if available
if 'volume_surge_threshold' in candles_df.columns:
    fig.add_trace(
        go.Scatter(
            x=candles_df['timestamp'],
            y=candles_df['volume_surge_threshold'],
            name=f'Volume Surge ({config.volume_surge})',
            line=dict(color='red', width=1, dash='dash')
        ),
        row=2, col=1
    )

# Add RSI if available
if 'rsi' in candles_df.columns:
    fig.add_trace(
        go.Scatter(
            x=candles_df['timestamp'],
            y=candles_df['rsi'],
            name=f'RSI ({config.rsi_period})',
            line=dict(color='purple')
        ),
        row=3, col=1
    )
    
    # Add RSI overbought/oversold lines
    fig.add_shape(
        type="line", line=dict(dash='dash', width=1, color="red"),
        x0=candles_df['timestamp'].iloc[0], x1=candles_df['timestamp'].iloc[-1], y0=70, y1=70,
        row=3, col=1
    )
    fig.add_shape(
        type="line", line=dict(dash='dash', width=1, color="green"),
        x0=candles_df['timestamp'].iloc[0], x1=candles_df['timestamp'].iloc[-1], y0=30, y1=30,
        row=3, col=1
    )

# Update layout
fig.update_layout(
    title=f'ESP Fibonacci Pivot Strategy Backtest: {config.trading_pair}',
    xaxis_rangeslider_visible=False,
    height=900,
    width=1200,
    template='plotly_dark',
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)

# Update y-axis for RSI
fig.update_yaxes(title_text="RSI", range=[0, 100], row=3, col=1)
fig.update_yaxes(title_text="Volume", row=2, col=1)
fig.update_yaxes(title_text="Price", row=1, col=1)

# Show the figure
fig.show()
```

### Backtesting Analysis

### Scatter of PNL per Trade
This bar chart illustrates the PNL for each individual trade. Positive PNLs are shown in green and negative PNLs in red, providing a clear view of profitable vs. unprofitable trades.


```python
import plotly.express as px

# Create a new column for profitability
executors_df['profitable'] = executors_df['net_pnl_quote'] > 0

# Create the scatter plot
fig = px.scatter(
    executors_df,
    x="timestamp",
    y='net_pnl_quote',
    title='PNL per Trade',
    color='profitable',
    color_discrete_map={True: 'green', False: 'red'},
    labels={'timestamp': 'Timestamp', 'net_pnl_quote': 'Net PNL (Quote)'},
    hover_data=['filled_amount_quote', 'side']
)

# Customize the layout
fig.update_layout(
    xaxis_title="Timestamp",
    yaxis_title="Net PNL (Quote)",
    legend_title="Profitable",
    font=dict(size=12, color="white"),
    showlegend=False,
    plot_bgcolor='rgba(0,0,0,0.8)',  # Dark background
    paper_bgcolor='rgba(0,0,0,0.8)',  # Dark background for the entire plot area
    xaxis=dict(gridcolor="gray"),
    yaxis=dict(gridcolor="gray")
)

# Add a horizontal line at y=0 to clearly separate profits and losses
fig.add_hline(y=0, line_dash="dash", line_color="lightgray")

# Show the plot
fig.show()
```

### Histogram of PNL Distribution
The histogram displays the distribution of PNL values across all trades. It helps in understanding the frequency and range of profit and loss outcomes.


```python
fig = px.histogram(executors_df, x='net_pnl_quote', title='PNL Distribution')
fig.show()
```
