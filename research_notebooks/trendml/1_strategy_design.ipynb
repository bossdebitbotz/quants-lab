```python
from data_sources.clob import ClobDataSource

# Get trading rules and candles
clob = ClobDataSource()
```

```python
thg_candles = await clob.get_candles_last_days(
    connector_name="okx",
    trading_pair="THG-USDT",
    interval="1m",
    days=5
)
btc_candles = await clob.get_candles_last_days(
    connector_name="okx",
    trading_pair="BTC-USDT",
    interval="1m",
    days=5
)
```

```python
btc_candles.plot("candles")
```

```python
thg_returns = thg_candles.data["close"].pct_change().cumsum() / thg_candles.data["close"].pct_change().std()
btc_returns = btc_candles.data["close"].pct_change().cumsum() / btc_candles.data["close"].pct_change().std()
```

```python
corr = thg_returns.corr(btc_returns)
```

```python
import plotly.graph_objects as go

# Plot the two lines
fig = go.Figure()
fig.add_trace(go.Scatter(x=thg_returns.index, y=thg_returns, mode="lines", name="THG"))
fig.add_trace(go.Scatter(x=btc_returns.index, y=btc_returns, mode="lines", name="BTC"))
fig.show()
```
